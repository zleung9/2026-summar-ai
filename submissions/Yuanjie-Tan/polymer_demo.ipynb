# 聚合物玻璃化转变温度（Tg）演示

这个 notebook 与 `polymer_self_demo.py` 配套，计算三种聚合物的**数量**
和**平均玻璃化转变温度（Tg）**，结果与脚本完全一致。

数据格式：聚合物名称 → Tg（摄氏度）

```python
polymers={"PE": -125, "PS": 100, "PET": 70} #载入数据
```

```python
def average_tg(polymers):
    return sum(polymers.values())/len(polymers)    #定义求平均函数
```

```python
def main():
    num_polymers=len(polymers) #计算聚合物数量
    print(f"聚合物数量：{num_polymers}")
    
    mean_tg=average_tg(polymers)  #计算聚合物Tg
    print(f"聚合物平均 Tg ：{mean_tg}")
```

```python
main()
```

## 结果

三种聚合物的平均 Tg = (−125 + 100 + 70) / 3 = **15.0 ℃**

| 项目 | 数值 |
|---|---|
| 聚合物数量 | **3** |
| 平均 Tg | **15.0 ℃** |

输出与 `polymer_self_demo.py` 完全一致，说明 notebook 与脚本使用的是
同一份数据和同一套计算逻辑。
