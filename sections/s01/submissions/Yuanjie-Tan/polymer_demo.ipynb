# Polymer Glass-Transition Temperature (Tg) Demo

This notebook accompanies `polymer_demo.py`. It calculates the **number of polymers**
in a tiny dataset and their **average glass-transition temperature (Tg)**,
which is a key thermal property used to classify polymer materials.

The dataset maps a polymer name to its Tg in degrees Celsius:

```python
polymers = {
    "PE": -125,   # polyethylene
    "PS": 100,    # polystyrene
    "PET": 70,    # polyethylene terephthalate
}
```

```python
def average_tg(polymer_data):
    """Return the mean glass-transition temperature of the given polymers."""
    return sum(polymer_data.values()) / len(polymer_data)
```

```python
num_polymers = len(polymers)
mean_tg = average_tg(polymers)

print(f"Number of polymers: {num_polymers}")
print(f"Average Tg: {mean_tg:.1f} C")
```

## Result

With the three polymers above, the **average Tg** is `( -125 + 100 + 70 ) / 3 = 15.0` C.
