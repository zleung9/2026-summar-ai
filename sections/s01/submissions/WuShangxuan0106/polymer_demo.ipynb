```python
polymers = {
    "PE": -125,
    "PS": 100,
    "PET": 70,
}
```

This code snippet contains known information about the polymer.

```python
polymer_count = len(polymers)
average_Tg = sum(polymers.values()) / len(polymers)
```

This code snippet calculates the count and average number of the polymer.

```python
print("polymer_count =", polymer_count)
print("average_Tg=", average_Tg)
```

output：

With the three polymers above, the **polymer_count** = 3 and the **average_Tg** is `( -125 + 100 + 70 ) / 3 = 15.0` C.
